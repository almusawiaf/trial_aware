```python
# diagnostic.py
import json
from matching_engine import ICD10Hierarchy

# Load hierarchy
h = ICD10Hierarchy("icd10_hierarchy.csv")
print(f"Hierarchy entries: {len(h.child_to_parent)}")

# Load trials
with open("data/10000_trials/structured_clinical_trials.json") as f:
    trials = json.load(f)

# Check code formats
sample_codes = set()
for trial in trials[:100]:
    for c in trial.get('criteria', []):
        if c.get('entity_code'):
            sample_codes.add(c['entity_code'])

print(f"Sample codes in trials: {list(sample_codes)[:20]}")
```
