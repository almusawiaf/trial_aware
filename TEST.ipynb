```python
import json, pandas as pd

# Load patient codes
diag_df = pd.read_parquet("./processed_data/diagnoses_clean.parquet")
patient_icd_codes = set(diag_df['ICD10_CODE'].dropna().unique())

# Load trial codes
with open("structured_clinical_trials.json") as f:
    trials = json.load(f)

trial_codes = set()
for t in trials:
    for c in t.get("inclusion_criteria", []) + t.get("exclusion_criteria", []):
        if "code" in c:
            trial_codes.add(c["code"])

overlap = patient_icd_codes.intersection(trial_codes)
print(f"Patient unique codes: {len(patient_icd_codes)}")
print(f"Trial unique codes: {len(trial_codes)}")
print(f"OVERLAP: {len(overlap)}")
```
