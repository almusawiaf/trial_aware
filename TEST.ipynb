```python
import json

with open("structured_clinical_trials.json") as f:
    trials = json.load(f)

# Print the first criterion of the first trial
first_trial = trials[0] if isinstance(trials, list) else list(trials.values())[0]
print("Keys in trial object:", list(first_trial.keys()))

inc = first_trial.get("inclusion_criteria", [])
if inc:
    print("First inclusion criterion structure:", json.dumps(inc[0], indent=2))
```
