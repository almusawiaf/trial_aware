```python
import torch
import torch.nn.functional as F

h_a = torch.load("processed_data/patient_embeddings_baseline.pt")
h_b = torch.load("processed_data/patient_embeddings.pt")

# Measure average pairwise cosine similarity across patients
sim_a = F.cosine_similarity(h_a.unsqueeze(1), h_a.unsqueeze(0), dim=-1).mean()
sim_b = F.cosine_similarity(h_b.unsqueeze(1), h_b.unsqueeze(0), dim=-1).mean()

print(f"Avg Pairwise Similarity Stage A: {sim_a:.4f}")
print(f"Avg Pairwise Similarity Stage B: {sim_b:.4f}")
```
